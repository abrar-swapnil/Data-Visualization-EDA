```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
student = pd.read_csv('sns_data (1).csv')
student.head()
```

# LINEPLOT USING SEABORN

```python
sns.lineplot(data=student, x='week', y='attendance_rate',errorbar=None)
```

```python
sns.lineplot(data=student, x='study_hours', y='test_score', errorbar=None, hue='gender')
plt.title('study hrs VS Score')
```

```python
sns.relplot(kind='line', data=student, x='study_hours', y='test_score', errorbar=None, hue='gender')
plt.title('study hrs VS Score')
```
