```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer

data = [18, 19, 20, 20, 21, 21, 22, 22, 23, 24, 25, 26, 27, 28, 30, 32, 35, 40, 45, 70]

df = pd.DataFrame(data,columns=['Age'])
```

```python
df.head()
```

```python
df['Age'].mean()
```

```python
sns.kdeplot(data=df, x='Age')
```

```python
age_mean = df['Age'].mean()
age_std = df['Age'].std()

df['Age_zscore'] = (df['Age'] - age_mean) / age_std

df.head(15)
```

```python
age_outliers = df[abs(df['Age_zscore']) > 3]
display(age_outliers)
```

```python
# removing outliers
df_no_outliers = df[abs(df['Age_zscore']) <= 3]
display(df_no_outliers) 
```
