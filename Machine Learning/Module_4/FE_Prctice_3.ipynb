```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd 
import sklearn 
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer

data = {'Salary' : [25, 28, 30, 32, 32, 33, 35, 35, 36, 38, 40, 42, 45, 48, 50, 55, 60, 65, 70, 80, 120, 150, 200, 250, 500]}
df = pd.DataFrame(data)
```

```python
df.head()
```

```python
salary_Q1 = df['Salary'].quantile(0.25)
salary_Q3 = df['Salary'].quantile(0.75)
salary_IQR = salary_Q3 - salary_Q1
lower_bound = max(0,salary_Q1 - 1.5 * salary_IQR)
upper_bound = salary_Q3 + 1.5 * salary_IQR
print(f"Lower Bound: {lower_bound}")
print(f"Upper Bound: {upper_bound}")
```

```python
salary_outliers = df[(df['Salary'] < lower_bound) | (df['Salary'] > upper_bound)]
print("Outliers in Salary:")
print(salary_outliers)
```

```python
new_df = df['Salary'].clip(lower = lower_bound, upper = upper_bound)
new_df = pd.DataFrame(data=new_df, columns=['Salary'])
```
