```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
```

```python


df = pd.read_csv('data.csv') 
df.head()
```

```python
X = df.drop('Target', axis=1)
y = df['Target']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

imputer = SimpleImputer(strategy='median')
imputer.fit(X_train[['Age', 'Salary']])
X_train[['Age', 'Salary']] = imputer.transform(X_train[['Age', 'Salary']])
X_test[['Age', 'Salary']] = imputer.transform(X_test[['Age', 'Salary']])
```

# Exam Cheat Sheet
* Numerical + Skewed/Outlier \rightarrow Always ⁠Median⁠
* Numerical + Normal/Bell \rightarrow Always ⁠Mean⁠
* Categorical + One Dominant (High %) \rightarrow Always ⁠Most_Frequent⁠
* Categorical + Balanced (No Dominant) / High Missing % \rightarrow Always ⁠Constant ('Missing')⁠
* Missing % > 50%-60% \rightarrow Direct ⁠Drop Column
